```python
from datasets import LMDBDataset
import torch
from torch.utils.data import DataLoader
from torchvision.utils import save_image, make_grid
from torchvision import models, transforms
```

```python
loader = DataLoader(LMDBDataset('./vqvae-code'), batch_size = 1)
```

```python
imgs = []
for n, i in enumerate(loader):
    imgs.append(transforms.ToPILImage()(i[0].to(dtype = torch.float32)))
    if  n == 14: break
import matplotlib.pyplot as plt

def display_images(images, cols=3):
    rows = (len(images) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(10, 10))
    
    for i, image in enumerate(images):
        ax = axes[i // cols, i % cols]
        ax.imshow(image, cmap='gray')
        ax.axis('off')
    
    plt.show()

display_images(imgs)
```
